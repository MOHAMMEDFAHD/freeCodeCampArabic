# Scikit-learn workflow contains the following steps:

1. Defining the problem (studying the business, defining the requirments)
2. Getting the data ready
3. Choosing the right estimator/algorithm (classification vs. regression)
4. Evaluate the model
5. Improve the model through hyperparameter experimentation
6. Deploy the model


```python
import numpy as np
```

```python
import pandas as pd

heart_disease = pd.read_csv("heart-disease.csv")
heart_disease
```

```python
# Creating a feature matrix (features/attributes/properatires/characteritics)
X = heart_disease.drop("target", axis=1)
# Creating the target variable (output vector)
y = heart_disease["target"]
```

```python
X
```

```python
y
```

## 2. Choosing the right estimator/model/algorithm

```python
from sklearn.ensemble import RandomForestClassifier
# we will keep all hyperparameters as default
clf = RandomForestClassifier()
```

```python
clf
```

## 3. Partioning our dataset into training and testing datasets

We will `80%` of the data for training and `20%` for testing

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
X_train # training feature matrix
```

```python
X_test # unseen, held-out and use later during evaluation/validation
```

```python
y_train # training labels are used during training
```

```python
y_test # testing lable used later for comparisons
```

## 4. Fitting the model on the data

```python
clf.fit(X_train, y_train) # here we are uisng fit method to tell the model go and find patterns/underlying structure
```

```python
clf.score(X_train, y_train)
```

## 5. Evaluate the model on the test data

```python
clf.score(X_test, y_test)
```

```python
clf.predict(np.array([0, 1, 2, 3, 4, 5])) # we have to pass the data in the same foramt as training data
```

```python
y_preds = clf.predict(X_test)
```

```python
y_preds
```

```python
y_test
```

# 6. Evluate the model further

```python
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

```python
print(classification_report(y_test, y_preds))
```

```python
confusion_matrix(y_test, y_preds)
```

```python
y_preds_train = clf.predict(X_train)
```

```python
y_preds_train
```

```python
accuracy_score(y_train, y_preds_train)
```

```python
accuracy_score(y_test, y_preds)
```

## 6. Let's implement hyperparameter tuning

```python
for i in range(10, 100, 10):
    print('Trying the model with ' + str(i) + " estimators")
    clf = RandomForestClassifier(n_estimators=i)
    clf.fit(X_train, y_train)
    y_preds = clf.predict(X_test)
    print(f"The model accuracy {accuracy_score(y_test, y_preds):.2f} %")
    print("")
```

```python
clf = RandomForestClassifier(n_estimators=90)
clf.fit(X_train, y_train)
clf.score(X_test, y_test)
```

```python
y_preds = clf.predict(X_test)
```

```python
y_preds
```

```python
print(classification_report(y_test, y_preds))
```

## Save the model and load it later

```python
import pickle
```

```python
pickle.dump(clf, open("my_model.pkl", "wb"))
```

```python
loaded_model = pickle.load(open("my_model.pkl", "rb"))
```
