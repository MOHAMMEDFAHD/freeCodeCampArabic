```python
# ============================================
# 🤗 Hugging Face Space – Generic Form + Inference Mode
# ============================================
# This template works with ANY form (text, sliders, files, etc.)
# and deploys it as an interactive Hugging Face Space with Inference enabled.
#
# Use case:
# - NLP / CV / Audio / Multimodal
# - Forms for prediction, classification, generation
# - Works locally AND on Hugging Face Spaces
#
#  How it works:
# 1. User fills a form
# 2. Data is passed to the inference function
# 3. Model runs (local or HF-hosted)
# 4. Result is returned instantly
#
# ============================================

import gradio as gr
from transformers import pipeline

# --------------------------------------------
#  LOAD MODEL (Inference Engine)
# --------------------------------------------
# Replace with ANY Hugging Face model
# Examples:
# - text-classification
# - text-generation
# - image-classification
# - speech-recognition
# --------------------------------------------

MODEL_NAME = "distilbert-base-uncased-finetuned-sst-2-english"

pipe = pipeline(
    task="text-classification",
    model=MODEL_NAME,
    device=-1  # CPU (HF Spaces compatible)
)

# --------------------------------------------
#  INFERENCE FUNCTION (Generic)
# --------------------------------------------
# This function receives form inputs
# and returns inference results
# --------------------------------------------

def run_inference(user_input):
    result = pipe(user_input)
    return {
        "label": result[0]["label"],
        "confidence": round(result[0]["score"], 4)
    }

# --------------------------------------------
#  FORM UI (Generic & Reusable)
# --------------------------------------------
# You can add:
# - Textboxes
# - Sliders
# - Dropdowns
# - File uploads
# --------------------------------------------

with gr.Blocks(
    title="🤗 Hugging Face Inference Space",
    theme=gr.themes.Soft()
) as demo:

    gr.Markdown(
        """
        ##  AI Inference Form
        **Submit input → Model runs → Get instant results**

         Powered by Hugging Face Transformers
         Inference Mode Enabled
        """
    )

    with gr.Row():
        user_text = gr.Textbox(
            label=" Input Text",
            placeholder="Type anything here...",
            lines=4
        )

    submit_btn = gr.Button(" Run Inference")

    output = gr.JSON(label=" Model Output")

    submit_btn.click(
        fn=run_inference,
        inputs=user_text,
        outputs=output
    )

# --------------------------------------------
#  LAUNCH (Inference Mode)
# --------------------------------------------
# IMPORTANT for Hugging Face Spaces:
# - server_name="0.0.0.0"
# - server_port=7860
# - share=False (HF handles public access)
# --------------------------------------------

demo.launch(
    server_name="0.0.0.0",
    server_port=7860,
    show_api=True
)

# ============================================
# DEPLOYMENT CHECKLIST (HF SPACE)
# ============================================
#  repository files:
# - app.py        ← this file
# - requirements.txt
#
#  requirements.txt example:
# gradio
# transformers
# torch
#
#  Space settings:
# - SDK: Gradio
# - Hardware: CPU (default)
# - Inference: ENABLED automatically
#
#  Done! Your form is now a live AI Space
# ============================================

```
