```python
import torch
import torch.nn.functional as F

# ---------------------------
# 1. Greedy Decoding (Argmax)
# ---------------------------
def greedy_decode(logits):
    """Return token index with maximum probability."""
    return torch.argmax(logits, dim=-1)


# ---------------------------
# 2. Top-k Sampling
# ---------------------------
def top_k_sampling(logits, k=50):
    """
    Keep only top-k logits, normalize them into probabilities,
    then sample one token from the filtered distribution.
    """
    values, indices = torch.topk(logits, k)
    probs = F.softmax(values, dim=-1)
    sampled = torch.multinomial(probs, 1)
    return indices[sampled]


# ---------------------------
# 3. Top-p (Nucleus) Sampling
# ---------------------------
def top_p_sampling(logits, p=0.9):
    """
    Sort tokens by probability, keep smallest number whose cumulative
    probability exceeds 'p', then sample one token.
    """
    sorted_logits, sorted_indices = torch.sort(logits, descending=True)
    sorted_probs = F.softmax(sorted_logits, dim=-1)
    cumulative_probs = sorted_probs.cumsum(dim=-1)

    # Mask tokens outside nucleus
    mask = cumulative_probs > p
    sorted_logits[mask] = float('-inf')

    # Sample from filtered logits
    filtered_probs = F.softmax(sorted_logits, dim=-1)
    sampled = torch.multinomial(filtered_probs, 1)

    # Return token index in original vocabulary
    return sorted_indices[sampled]


# ---------------------------
# 4. Temperature Sampling
# ---------------------------
def temperature_sampling(logits, temperature=1.0):
    """
    Scale logits by temperature before sampling.
    Lower temperature → sharper distribution.
    """
    scaled = logits / temperature
    probs = F.softmax(scaled, dim=-1)
    return torch.multinomial(probs, 1)


# ---------------------------
# 5. Random Sampling from Full Distribution
# ---------------------------
def random_sampling(logits):
    """
    Sample directly from softmax distribution without filtering.
    """
    probs = F.softmax(logits, dim=-1)
    return torch.multinomial(probs, 1)


# ---------------------------
# 6. Utility: Get Top-k Tokens for Display
# ---------------------------
def top_k_tokens(logits, tokenizer, k=10):
    values, indices = torch.topk(logits, k)
    return [(tokenizer.decode([idx]), float(values[i])) for i, idx in enumerate(indices)]


sentence = "Today I decided to go to the local mall and"
inputs = tokenizer(sentence, return_tensors="pt")
output = gpt2(**inputs)
logits = output.logits[0, -1]

print("Greedy:", tokenizer.decode([greedy_decode(logits)]))
print("Top-k:", tokenizer.decode(top_k_sampling(logits, k=10)))
print("Top-p:", tokenizer.decode(top_p_sampling(logits, p=0.9)))
print("Temp:", tokenizer.decode(temperature_sampling(logits, temperature=0.7)))
print("Random:", tokenizer.decode(random_sampling(logits)))
print(f"==============\n")

top10 = torch.topk(logits.softmax(dim=0), 10)
for value, index in zip(top10.values, top10.indices):
    print(f"{tokenizer.decode(index)} -- {value:.1%}")
```
